```python
features = ['valence', 'acousticness', 'danceability', 'energy', 'instrumentalness',
       'liveness', 'loudness', 'popularity', 'speechiness', 'tempo']

plt.figure(figsize = (12,10))
# plt.rcParams.update({'font.size': 12})
sns.heatmap(df[features].corr(), annot=True, fmt=".1f", linewidths=.5, cmap="crest")
```

**How to interpret it :**

The **darker** the color, the closer it is to **1.0** => *postitive correlation* (greater than or equal to 0.6)

The **lighter** the color, the closer it is to **-1.0** => *negative correlation* (less than or equal to -0.6)
